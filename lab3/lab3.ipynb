## III – THỰC HÀNH CÁC THAO TÁC TRÊN CỘT DỮ LIỆU
Chúng ta sử dụng bộ dữ liệu mẫu `sample_data.csv` gồm thông tin nhân sự/khách hàng tương ứng với phần hướng dẫn thực hành trên Google Colab.

```python
# Đọc bộ dữ liệu mẫu
df = pd.read_csv(f'{DATASET_BASE_URL}/sample_data.csv')
df.head()
```

### 1. Chuẩn hóa viết thường, in hoa, viết hoa đầu từ tên cột

```python
# Thay đổi tiêu đề các dòng thành chữ thường
df.columns.str.lower()
```

```python
# Thay đổi thành chữ in hoa
df.columns.str.upper()
```

```python
# Ký tự đầu in hoa
df.columns.str.title()
```

### 2. Thay thế ký tự khoảng trắng bằng dấu gạch dưới `_`

```python
# Thay thế khoảng trắng bằng dấu _
df.columns.str.replace(' ', '_')
```

### 3. Đổi tên cột (`rename` và gán trực tiếp qua `columns.values`)

```python
# Thay đổi tên cột Age thành Date of Birth
df = df.rename(columns={'Age': 'Date of Birth'})
df.head()
```

```python
# Đổi tên cột thứ 8 (chỉ số 7) thành Email Address
df.columns.values[7] = 'Email Address'
df.columns
```

### 4. Lọc chọn các cột theo điều kiện

```python
# Chọn tất cả trừ điều kiện (!= 'SALARY')
df.columns[df.columns != 'SALARY']
```

```python
# Chọn tất cả trừ điều kiện dùng phủ định isin
df.loc[:, ~df.columns.isin(['SALARY'])].columns
```

```python
# Chọn tên cột chứa ký tự trong điều kiện (df.filter like)
df.filter(like='STREET').columns
```

```python
# Chọn tên cột với các ký tự cuối trong điều kiện (endswith)
df.loc[:, df.columns.str.endswith('name') | df.columns.str.endswith('Name')].columns
```

```python
# Hiện tên cột với ký tự cuối trong điều kiện bằng bộ lọc regex
df.filter(regex='(?i)name$', axis=1).columns
```

```python
# Hiện một nhóm tên cột theo chỉ mục slice
df.columns.values[0:4]
```

## IV – BÀI TẬP VỀ NHÀ: LÀM SẠCH BỘ DỮ LIỆU BẤT ĐỘNG SẢN (`property_data.csv`)

Bộ dữ liệu trong repository: `https://raw.githubusercontent.com/shizziio/uit-DS200.G11.CN2.TTNT/main/datasets/lab3/property_data.csv`  
Bao gồm các thuộc tính:
- `PID`: Mã định danh bất động sản
- `ST_NUM`: Số nhà (Street Number)
- `ST_NAME`: Tên đường (Street Name)
- `OWN_OCCUPIED`: Đang có chủ ở hay không (Y/N)
- `NUM_BEDROOMS`: Số phòng ngủ
- `NUM_BATH`: Số phòng tắm
- `SQ_FT`: Diện tích (feet vuông)

### 2.1. Đọc tệp csv vào Dataframe
*Lưu ý:* Trong thực tế, dữ liệu khuyết thiếu không chỉ xuất hiện dưới dạng ô trống hoặc `NaN`, mà còn có `n/a`, `na`, `--`. Chúng ta truyền danh sách `na_values` để Pandas nhận diện đồng nhất tất cả thành `NaN`.

```python
# Đọc tệp property_data.csv có xử lý các định dạng missing đặc thù
missing_values = ['n/a', 'na', '--']
df_prop = pd.read_csv(f'{DATASET_BASE_URL}/property_data.csv', na_values=missing_values)
df_prop
```

### 2.2. Tìm trong Dataframe những dữ liệu bị thiếu
Sử dụng `isnull()` để kiểm tra ma trận boolean và đếm số lượng giá trị thiếu theo từng cột.

```python
# Kiểm tra dữ liệu bị thiếu dạng bảng boolean
df_prop.isnull()
```

```python
# Thống kê số lượng ô bị thiếu trên từng cột
df_prop.isnull().sum()
```

### 2.3. Xem cột `NUM_BEDROOMS` và tìm những giá trị `NaN`

```python
# Xem giá trị cột NUM_BEDROOMS
print(df_prop['NUM_BEDROOMS'])

# Lọc và hiển thị các dòng có NUM_BEDROOMS là NaN
df_prop[df_prop['NUM_BEDROOMS'].isnull()]
```

### 2.4. Tổng giá trị còn thiếu trong Dataframe
Sử dụng `df.isnull().sum().sum()` để tính tổng toàn bộ ô trống/khuyết trong DataFrame.

```python
total_missing = df_prop.isnull().sum().sum()
print(f"Tổng số giá trị còn thiếu trong toàn bộ Dataframe là: {total_missing}")
```

### 2.5a. Điền giá trị 126 vào ô bị thiếu trong cột `ST_NUM`

```python
# Điền giá trị 126 vào các ô thiếu trong ST_NUM
df_prop['ST_NUM'] = df_prop['ST_NUM'].fillna(126)
df_prop[['PID', 'ST_NUM', 'ST_NAME']]
```

### 2.5b. Thay thế missing values trong cột `NUM_BEDROOMS` bằng giá trị trung vị (median)

```python
# Tính giá trị trung vị của cột NUM_BEDROOMS
median_bedrooms = df_prop['NUM_BEDROOMS'].median()
print(f"Giá trị median của NUM_BEDROOMS: {median_bedrooms}")

# Thay thế các ô NaN bằng median
df_prop['NUM_BEDROOMS'] = df_prop['NUM_BEDROOMS'].fillna(median_bedrooms)
df_prop[['PID', 'NUM_BEDROOMS']]
```
