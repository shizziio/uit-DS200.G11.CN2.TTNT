## I – LÝ THUYẾT VỀ MISSING DATA
- **Missing data là gì?** Dữ liệu bị thiếu, hiển thị dưới dạng `NaN`, `NaT`, `None`, `Null`, `N/A`, v.v.
- **Nguyên nhân xuất hiện:**
  1. Người dùng quên điền hoặc bỏ sót trường thông tin.
  2. Dữ liệu bị thất thoát trong quá trình chuyển giao thủ công từ cơ sở dữ liệu cũ sang hệ thống mới.
  3. Lỗi chương trình thu thập, lỗi đường truyền hoặc sự cố cảm biến đo đạc.
  4. Trùng hợp ngẫu nhiên.
- **Phân loại Missing Data:**
  1. **MCAR (Missing Completely at Random):** Dữ liệu thiếu hoàn toàn ngẫu nhiên. Xác suất dữ liệu bị thiếu không phụ thuộc vào giá trị của bất kỳ biến nào (kể cả biến đang xét lẫn các biến khác).
  2. **MAR (Missing at Random):** Dữ liệu khuyết ngẫu nhiên có điều kiện. Xác suất dữ liệu bị thiếu phụ thuộc vào các biến đã quan sát được trong tập dữ liệu.
  3. **MNAR (Missing Not at Random):** Dữ liệu khuyết không ngẫu nhiên. Xác suất dữ liệu bị thiếu phụ thuộc trực tiếp vào chính giá trị của biến đó.
- **Pandas handling:** Thư viện Pandas xem `None` và `NaN` (Not a Number từ NumPy) về cơ bản có thể hoán đổi cho nhau để biểu thị các giá trị còn thiếu, đồng thời cung cấp bộ công cụ mạnh mẽ: `isnull()`, `notnull()`, `dropna()`, `fillna()`, `replace()`, `interpolate()`.

Sử dụng hàm `isnull()` để kiểm tra từng phần tử trong DataFrame và trả về ma trận boolean (`True` nếu là giá trị khuyết, `False` nếu có giá trị).


```python
# importing pandas as pd
import pandas as pd
# importing numpy as np
import numpy as np

# tạo dictionary of lists
dict = {
    'Điểm dòng 1': [100, 90, np.nan, 95],
    'Điểm dòng 2': [30, 45, 56, np.nan],
    'Điểm dòng 3': [np.nan, 40, 80, 98]
}

# Tạo một dataframe từ list
df = pd.DataFrame(dict)

# Dùng hàm isnull()
print(df.isnull())
```

### 3. Điền các giá trị còn thiếu bằng cách sử dụng fillna(), replace() và interpolate()

#### 3.1. Sử dụng hàm fillna() với giá trị hằng số (0)
Điền tất cả các vị trí có giá trị thiếu (`NaN`) bằng số 0.


```python
# importing pandas as pd
import pandas as pd
# importing numpy as np
import numpy as np

# dictionary of lists
dict = {
    'Điểm dòng 1': [100, 90, np.nan, 95],
    'Điểm dòng 2': [30, 45, 56, np.nan],
    'Điểm dòng 3': [np.nan, 40, 80, 98]
}

# Tạo một dataframe từ dictionary
df = pd.DataFrame(dict)

# Tìm những giá trị còn thiếu với fillna()
print(df.fillna(0))
```

#### 3.2. Sử dụng hàm fillna() với method='pad' (hoặc ffill)
Điền giá trị còn thiếu bằng giá trị hợp lệ liền trước nó trong cùng một cột.


```python
# importing pandas as pd
import pandas as pd
# importing numpy as np
import numpy as np

# dictionary of lists
dict = {
    'Điểm dòng 1': [100, 90, np.nan, 95],
    'Điểm dòng 2': [30, 45, 56, np.nan],
    'Điểm dòng 3': [np.nan, 40, 80, 98]
}

# Tạo một dataframe từ dictionary
df = pd.DataFrame(dict)

# điền một giá trị còn thiếu với những giá trị trước đó
print(df.ffill())
```

#### 3.3. Sử dụng hàm replace()
Thay thế các giá trị `NaN` trong toàn bộ DataFrame của tập dữ liệu `employees.csv` bằng một giá trị đại diện (ví dụ `-88`).


```python
# importing pandas package
import pandas as pd
import numpy as np

data = pd.read_csv("https://raw.githubusercontent.com/shizziio/uit-DS200.G11.CN2.TTNT/main/datasets/lab2/employees.csv")

# Thay thế NaN trong dataframe với giá trị -88
data_new = data.replace(to_replace=np.nan, value=-88)
print(data_new)
```

#### 3.4. Sử dụng hàm interpolate() để điền các giá trị còn thiếu bằng phương pháp tuyến tính (Linear Interpolation)
Nội suy tuyến tính giúp ước lượng giá trị khuyết thiếu dựa trên độ dốc và khoảng cách giữa hai mốc giá trị thực tế trước và sau nó.


```python
import pandas as pd

df = pd.DataFrame({
    "A": [12, 4, 5, None, 1],
    "B": [None, 2, 54, 3, None],
    "C": [20, 16, None, 3, 8],
    "D": [14, 3, None, None, 6]
})
print(df)
print("================================")
data_new = df.interpolate(method='linear', limit_direction='forward')
print(data_new)
```
